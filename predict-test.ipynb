For our web service, the churn prediction service cannot be accessed via a browser as it sends a `GET` request : <br>`127.0.0.1 - - [30/Jun/2025 13:51:40] "GET /predict HTTP/1.1" 405`

Hence we need to send a `POST` request

```python
import requests
```

```python
url = 'http://localhost:9696/predict'
```

```python
# should be in json
customer = {
    "gender": "female",
    "seniorcitizen": 0,
    "partner": "yes",
    "dependents": "yes",
    "phoneservice": "yes",
    "multiplelines": "yes",
    "internetservice": "fiber_optic",
    "onlinesecurity": "yes",
    "onlinebackup": "no",
    "deviceprotection": "yes",
    "techsupport": "no",
    "streamingtv": "yes",
    "streamingmovies": "yes",
    "contract": "month-to-month",
    "paperlessbilling": "yes",
    "paymentmethod": "electronic_check",
    "tenure": 2,
    "monthlycharges": 104.2,
    "totalcharges": 29.85
 }
```

```python
# sending this customer to POST request - flask app should be running
requests.post(url, json=customer)
```

`<Response [200]>` means it was successful -> `127.0.0.1 - - [30/Jun/2025 14:16:58] "POST /predict HTTP/1.1" 200`

```python
# converting request content and turning it to dict
response = requests.post(url, json=customer).json()
response
```

```python
# usage by marketing service
if response['Churn'] == True:
    print("sending promo email to %s" %('customer-xyz'))
```
